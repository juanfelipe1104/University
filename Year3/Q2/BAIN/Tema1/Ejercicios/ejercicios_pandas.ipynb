## Ejercicio 3
Leer el archivo CSV `datos_ejemplo.csv`, parsear la columna `birthdate` como fechas y establecerla como índice del DataFrame.

```python
df = pd.read_csv('datos_ejemplo.csv')
df['birthdate'] = pd.to_datetime(df['birthdate'], format='%Y-%m-%d')
# df.index = df['birthdate']
display(df.head())
```

# Ejercicios de Selección, Filtrado y Creación de Nuevos DataFrames
## Ejercicio 1
Seleccionar y mostrar solo las filas donde el apellido sea `Miller`.

```python
miller_rows = df[df['lastname'] == 'Miller']
display(miller_rows)
```

## Ejercicio 2
Seleccionar y mostrar solo las filas donde la fecha de nacimiento sea posterior a `1990-01-01`.

```python
after_1990 = df[df['birthdate'] > pd.Timestamp('1990-01-01')]
display(after_1990)
```

## Ejercicio 3
Crear un nuevo DataFrame con una columna adicional `fullname` que contenga el nombre completo (`firstname` + `lastname`).

```python
df['full_name'] = df['firstname'] + ' ' + df['lastname']
display(df.head())
```

## Ejercicio 4
Seleccionar y mostrar solo las filas donde el nombre sea `Ellen`.

```python
ellen_rows = df[df['firstname'] == 'Ellen']
display(ellen_rows)
```

## Ejercicio 5
Seleccionar y mostrar solo las filas donde el correo electrónico contenga `gmail.com`.

```python
gmail_rows = df[df['email'].str.contains('gmail.com', na=False)]
display(gmail_rows)
```

## Ejercicio 6
Crear un nuevo DataFrame con una columna adicional `age` que contenga la edad calculada a partir de la fecha de nacimiento.

```python
today = pd.Timestamp('today')
before_birthday = (
    (df['birthdate'].dt.month > today.month) |
    ((df['birthdate'].dt.month == today.month) & (df['birthdate'].dt.day > today.day))
)

df['age'] = today.year - df['birthdate'].dt.year - before_birthday.astype(int)
display(df.head())
```

## Ejercicio 7
Seleccionar y mostrar solo las filas donde la edad sea mayor a 30 años.

```python
over_30_years = df[df['age'] > 30]
display(over_30_years)
```

## Ejercicio 8
Seleccionar y mostrar solo las filas donde el nombre completo contenga `Miller`.

```python
miller_rows = df[df['full_name'].str.contains('Miller', na=False)]
display(miller_rows)
```

## Ejercicio 9
Crear un nuevo DataFrame con una columna adicional `domain` que contenga el dominio del correo electrónico.

```python
df['domain'] = df['email'].str.split('@').str[1]
display(df.head())
```

## Ejercicio 10
Seleccionar y mostrar solo las filas donde el dominio del correo electrónico sea `hotmail.com`.

```python
hotmail_rows = df[df['domain'] == 'hotmail.com']
display(hotmail_rows)
```

# Ejercicios de Agrupación, Ordenación y Unión
## Ejercicio 1
Agrupar por apellido y contar el número de ocurrencias de cada apellido.

```python
counts_by_lastname = df.groupby('lastname').size().sort_values(ascending=False)
display(counts_by_lastname)
```

## Ejercicio 2
Ordenar el DataFrame por fecha de nacimiento en orden ascendente.

```python
sorted_by_birthdate = df.sort_values('birthdate', ascending=True)
display(sorted_by_birthdate)
```

## Ejercicio 3
Unir dos DataFrames creados a partir del archivo CSV original.

```python
df_test = pd.read_csv('datos_ejemplo.csv')

df1 = df_test[['id', 'firstname', 'lastname']].copy()
df2 = df_test[['id', 'email', 'birthdate']].copy()

merged = pd.merge(df1, df2, on='id', how='inner')
display(merged.head())
```

## Ejercicio 4
Agrupar por dominio del correo electrónico y contar el número de ocurrencias de cada dominio.

```python
counts_by_domain = df.groupby('domain').size().sort_values(ascending=False)
display(counts_by_domain)
```

## Ejercicio 5
Ordenar el DataFrame por nombre completo en orden descendente.

```python
sorted_by_fullname_desc = df.sort_values('full_name', ascending=False)
display(sorted_by_fullname_desc)
```

## Ejercicio 6
Unir dos DataFrames utilizando la columna `id` como clave.

```python
df_test = pd.read_csv('datos_ejemplo.csv')

left = df_test[['id', 'firstname', 'lastname']].copy()
right = df_test[['id', 'email']].copy()

joined_on_id = left.merge(right, on='id', how='inner')
display(joined_on_id.head())
```

## Ejercicio 7
Agrupar por año de nacimiento y contar el número de ocurrencias de cada año.

```python
df['birth_year'] = df['birthdate'].dt.year
counts_by_year = df.groupby('birth_year').size().sort_values(ascending=False)
display(counts_by_year)
```

## Ejercicio 8
Ordenar el DataFrame por apellido en orden ascendente y luego por nombre en orden descendente.

```python
sorted_multi = df.sort_values(['lastname', 'firstname'], ascending=[True, False])
display(sorted_multi)
```

## Ejercicio 9
Unir dos DataFrames utilizando la columna `email` como clave.

```python
df_test = pd.read_csv('datos_ejemplo.csv')

a = df_test[['email', 'firstname', 'lastname']].copy()
a['fullname'] = a['firstname'] + ' ' + a['lastname']
a = a[['email', 'fullname']]

b = df_test[['email', 'birthdate']].copy()

joined_on_email = a.merge(b, on='email', how='inner')
display(joined_on_email.head())
```
